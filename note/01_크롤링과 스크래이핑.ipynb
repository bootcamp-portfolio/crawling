###스크래이핑,크롤링

####스크레이핑(scraping)
-웹사이트에 있는 특정 정보를 추출하는 기술
-웹에 공개된 정보는 html형식
-html 데이터를 획득하여 데이터 베이스에 저장하려면 가공기술 필요
-광고 등의 불 필요한 정보를 제공하고 필요한 정보를 가져오려면 사이트의 구조를 분석 가능해야함.

####크롤링(crawling)
-프로그램이 웹사이트를 정기적으로 정보를 추출하는 기술

->api #로보츠 정책 robots.txt

### requests
#### 응답코드
-1xx:요청은 받았고 작업 진행중
-2xx:사용자의 요청이 성공적으로 수행됨(정상)
-3xx:요청은 완료되었으나 리다이렉션 필요 (다른서버에 연결중)
-4xx:(404 낫파운드) 사용자의 요청이 잘못됐다
-5xx:서버에 오류 발생

```python
#request 응답코드
import requests
URL="https://www.naver.com"
response=requests.get(URL)
print(response.status_code)
print(response.text)
```

```python
#데이터 다운로드
import urllib.request as req

#File Name
fileName='mov01.jpg'

#url과 다운 받을 경로 지정하기
url="https://zeushahn.github.io/Test?images/"+fileName
saveName="../Data/"+fileName

#다운로드
req.urlretrieve(url,saveName)
print("ok!")
```

```python
#ex
fileLists=['mov02.jpg','mov03.jpg','mov04.jpg']
for fileName in fileLists:
   url="https://zeushahn.github.io/Test/images/"+fileName
   saveName="../Data/"+fileName
   req.urlretrieve(url,saveName)
   print(f"{fileName} is downloaded.")
```
