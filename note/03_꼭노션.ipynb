```python
from bs4 import BeautifulSoup
import urllib.request as req
import urllib.parse
import time

# 저장소 정의
title = []
contents = []

# html 가져오기

url = 'https://news.daum.net/tech'
res = req.urlopen(url)

# html 분석하기

soup = BeautifulSoup(res, 'html.parser')
# 데이터 추출하기

a_list = soup.select(".item_newsbasic")
a_list = a_list[:9]

for a in a_list:
    # print(a.attrs['href'], ":", end="")
    # print(urllib.parse.unquote(a.attrs['data-title']))
    title.append(urllib.parse.unquote(a.attrs['data-title']))

    sub_url = a.attrs['href']
    sub_res = req.urlopen(sub_url)
    sub_soup = BeautifulSoup(sub_res, 'html.parser')
    sub_a_list = sub_soup.select('#mArticle > div.news_view.fs_type1 > div.article_view > section > p')
    # print(sub_a_list)

    temp_contents = ""
    for sub_a in sub_a_list:
        temp_contents += sub_a.text

    contents.append(temp_contents)

    time.sleep(1)

    
for t, c in zip(title, contents):
    print(f'제목 : {t}')
    print('-'*20)
    print(c)
    print('-'*40)

```
