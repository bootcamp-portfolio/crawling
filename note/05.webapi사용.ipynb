```python
import requests
import json
import time

def k2c(k):
    return round(k - 273.15, 2)

apiKey = "a78d950e59a3d526a61b92be517488a7"
api = "http://api.openweathermap.org/data/2.5/weather?q={city}&APPID={key}"
cities = ['Seoul', 'Tokyo', 'New York']

for city in cities:
    
    url = api.format(city=city, key=apiKey)
    
    r = requests.get(url)
    data = json.loads(r.text)

    if data.get('cod') != 200:
        print(f"Error fetching {city}: {data.get('message')}")
        continue

    print(f"+도시 - {data['name']} ")
    print("|날씨 -", data['weather'][0]['description'])
    print("|최저기온 -", k2c(data['main']['temp_min']), "°C")
    print("|최고기온 -", k2c(data['main']['temp_max']), "°C")
    print("|습도 -", data['main']['humidity'], "%")
    print("|기압 -", data['main']['pressure'], "hPa")
    print("|풍향 -", data['wind']['deg'], "°")
    print("|풍속 -", data['wind']['speed'], "m/s")
    print("\n")
```
